# Agribusiness Data Analysis and Visualization
**Week 3 Internship Task**

> Values are illustrative sample data, not real farm or market records.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

df = pd.read_csv('../data/agribusiness_sample_data.csv')
df.head()
```

## 1. Data checks

```python
print('Shape:', df.shape)
print('\nMissing values:\n', df.isna().sum())
print('\nSummary:\n', df.describe(numeric_only=True))
```

## 2. Crop yield trend

```python
plt.figure(figsize=(8,4))
plt.plot(df.month, df.wheat_yield_t_ha, marker='o', label='Wheat')
plt.plot(df.month, df.vegetable_yield_t_ha, marker='o', label='Vegetables')
plt.title('Illustrative crop-yield trend')
plt.xlabel('Month'); plt.ylabel('Yield (tonnes per hectare)')
plt.grid(alpha=.25); plt.legend(); plt.tight_layout()
plt.savefig('../visualizations/yield_trend.png', dpi=160)
plt.show()
```

## 3. Operating cost comparison

```python
costs = pd.Series({'Fertilizer':26500, 'Labour':22000, 'Seeds':18000, 'Irrigation':12500, 'Transport':11000})
ax = costs.plot(kind='bar', figsize=(8,4), title='Illustrative farm operating costs')
ax.set_xlabel('Cost category'); ax.set_ylabel('Cost (INR)'); ax.grid(axis='y', alpha=.25)
plt.xticks(rotation=20); plt.tight_layout()
plt.savefig('../visualizations/cost_breakdown.png', dpi=160)
plt.show()
```

## 4. Market price and supply

```python
fig, ax1 = plt.subplots(figsize=(8,4))
ax1.plot(df.month, df.market_price_inr_kg, marker='o', label='Market price')
ax1.set_xlabel('Month'); ax1.set_ylabel('Illustrative price (INR/kg)'); ax1.grid(alpha=.25)
ax2 = ax1.twinx(); ax2.plot(df.month, df.market_supply_index, marker='s', linestyle='--', label='Supply index')
ax2.set_ylabel('Supply index')
lines = ax1.get_lines() + ax2.get_lines(); ax1.legend(lines, [x.get_label() for x in lines], loc='upper left')
plt.title('Illustrative market price and supply'); fig.tight_layout()
plt.savefig('../visualizations/market_price_supply.png', dpi=160); plt.show()
```

## 5. Fertilizer spending and yield

```python
fertilizer_index = [14,18,19,22,24,27,29,31]
yield_values = [2.7,3.0,3.1,3.3,3.5,3.7,3.8,4.0]
plt.figure(figsize=(7,4)); plt.scatter(fertilizer_index, yield_values)
plt.title('Illustrative fertilizer-spending index vs yield')
plt.xlabel('Fertilizer spending index'); plt.ylabel('Yield (tonnes per hectare)')
plt.grid(alpha=.25); plt.tight_layout()
plt.savefig('../visualizations/fertilizer_yield_scatter.png', dpi=160); plt.show()
```

## 6. Findings and limitations
The sample yield series varies by month. Cost categories can be reviewed, but a high cost is not automatically wasteful. The price/supply pattern does not prove causation. The dataset is small and illustrative, so it cannot support statistical-significance or causal claims. Replace it with verified observations before making decisions.
